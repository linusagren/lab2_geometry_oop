# Manual cube tests

## Valid surface area and volume

```python
from cube import Cube

c1 = Cube(3)
print("Verify cube initialization values:")
print(f"{c1.edge_length = }") # Expected: 3
print(f"{c1.surface_area = }") # Expected: 54
print(f"{c1.volume = }") # Expected: 27

c1.edge_length = 4
print("Verify cube values after changing edge length:")
print(f"{c1.edge_length = }") # Expected: 4
print(f"{c1.surface_area = }") # Expected: 96
print(f"{c1.volume = }") # Expected: 64
```

## Invalid edge length values

```python
c2 = Cube(3)

# String
try:
    c2.edge_length = "..."
except TypeError as err:
    print(err)
    print(f"{c2.edge_length = } | Expected: 3")

# Boolean
try:
    c2.edge_length = True
except TypeError as err:
    print(err)
    print(f"{c2.edge_length = } | Expected: 3")

# Zero
try:
    c2.edge_length = 0
except ValueError as err:
    print(err)
    print(f"{c2.edge_length = } | Expected: 3")

# Negative
try:
    c2.edge_length = -3
except ValueError as err:
    print(err)
    print(f"{c2.edge_length = } | Expected: 3")
```

## \_\_str\_\_ and \_\_repr\_\_

```python
print(str(c2))
print(repr(c2))
```

## Translate tests

### Valid translate

```python
c3 = Cube(3, 3, 3, 5)
print("Initial values:")
print(f"{c3.edge_length = }")
print(f"{c3.x = }")
print(f"{c3.y = }")
print(f"{c3.z = }")

print("\nAfter translating x and y:")
c3.translate(1, 1)
print(f"{c3.edge_length = } | Expected: 3")
print(f"{c3.x = } | Expected: 4")
print(f"{c3.y = } | Expected: 4")
print(f"{c3.z = } | Expected: 5")
print(str(c3))
print(repr(c3))

print("\nAfter translating x, y and z:")
c3.translate(-2, 3, 3)
print(f"{c3.edge_length = } | Expected: 3")
print(f"{c3.x = } | Expected: 2")
print(f"{c3.y = } | Expected: 7")
print(f"{c3.z = } | Expected: 8")
print(str(c3))
print(repr(c3))
```

## Operator overload tests

```python
c1 = Cube(4)
c2 = Cube(4)
c3 = Cube(7)

# __eq__
print(f"{c1.volume == c2.volume = }") # Expected: True
print(f"{c1.volume == c3.volume = }") # Expected: False

# __lt__
print(f"{c1.volume < c3.volume = }") # Expected: True
print(f"{c3.volume < c1.volume = }") # Expected: False

# __gt__
print(f"{c1.volume > c2.volume = }") # Expected: False
print(f"{c3.volume > c2.volume = }") # Expected: True

# __ge__
print(f"{c1.volume >= c2.volume = }") # Expected: True
print(f"{c3.volume >= c2.volume = }") # Expected: True
print(f"{c1.volume >= c3.volume = }") # Expected: False

# __le__
print(f"{c1.volume <= c2.volume = }") # Expected: True
print(f"{c3.volume <= c2.volume = }") # Expected: False
print(f"{c1.volume <= c3.volume = }") # Expected: True
```
