## Operator overload tests with separate subclasses

```python
from circle import Circle
from rectangle import Rectangle

r1 = Rectangle(4, 4)
c1 = Circle(5)

print(f"{r1.area = }")
print(f"{c1.area = }")
print()
print(f"{r1 == c1 = }") # Expected: False
print(f"{r1 < c1 = }") # Expected: True
print(f"{r1 > c1 = }") # Expected: False
print(f"{r1 <= c1 = }") # Expected: True
print(f"{r1 >= c1 = }") # Expected: False
```
