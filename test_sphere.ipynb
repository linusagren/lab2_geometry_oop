# Sphere tests

## Valid and invalid init

```python
from sphere import Sphere

# Valid
sph1 = Sphere(5)
sph1 = Sphere(5, 5, -2, 0)

# Invalid
try:
    sph2 = Sphere(0)
except ValueError as err:
    print(err)

try:
    sph2 = Sphere(-3)
except ValueError as err:
    print(err)

try:
    sph2 = Sphere("...")
except TypeError as err:
    print(err)

try:
    sph2 = Sphere(True)
except TypeError as err:
    print(err)

try:
    sph3 = Sphere(5, "...", -2, 0)
except TypeError as err:
    print(err)

try:
    sph3 = Sphere(5, -2, True, 0)
except TypeError as err:
    print(err)
```

## Getter and Setter: volume & surface area

```python
sph1 = Sphere(5)

print(f"{sph1.volume = }")
print(f"{sph1.surface_area = }")

# Invalid Assignment
try:
    sph1.volume = 32
except AttributeError as err:
    print(err)

try:
    sph1.surface_area = 521
except AttributeError as err:
    print(err)
```

## Getter and Setter: radius

```python
sph1 = Sphere(8)

print(f"{sph1.radius = }, Coordinates: ({sph1.x}, {sph1.y}, {sph1.z})") # 8 | (0, 0, 0)

sph1.radius = 3
print(f"{sph1.radius = }, Coordinates: ({sph1.x}, {sph1.y}, {sph1.z})") # 3 | (0, 0, 0)
```

## Circumference

```python
sph1 = Sphere(5)

print(f"{sph1.circumference = }") # Expected: 31.4159

try:
    sph1.circumference = 22
except AttributeError as err:
    print(err)

print(f"{sph1.circumference = }") # Expected: 31.4159
```

## Translate tests

```python
# Valid
sph1 = Sphere(5, 0, -5, 3)
print(f"{sph1 = }")

sph1.translate(4, 2, -4) 
print(f"{sph1 = }") # Expected: 5, 4, -3, -1

# Invalid
try:
    sph1.translate("...", 4, 5)
except TypeError as err:
    print(f"{sph1 = } | {err}") # Expected: sph1 stays 5, 4, -3, -1 | TypeError
```

## \_\_str\_\_ and \_\_repr\_\_

```python
sph1 = Sphere(5)
sph2 = Sphere(2, 5, 1, -5)

print(str(sph1)) # A sphere with radius 5 on position (0, 0, 0).
print(str(sph2)) # A sphere with radius 2 on position (5, 1, -5).

print(repr(sph1)) # Sphere(5, 0, 0, 0)
print(repr(sph2)) # Sphere(2, 5, 1, -5)
```
