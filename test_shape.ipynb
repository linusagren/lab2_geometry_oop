# Shape Manual Tests

## Validate init

```python
from shape import Shape

s1 = Shape(4, 7)
s2 = Shape()

print(f"{s1.x = }") # Expected: 4
print(f"{s1.y = }") # Expected: 7
print(f"{s2.x = }") # Expected: 0
print(f"{s2.y = }") # Expected: 0

try:
    s3 = Shape(True, 5)
except TypeError as err:
    print(f"Boolean: {err}")

try:
    s3 = Shape("...", 5)
except TypeError as err:
    print(f"String: {err}")
```

## Validate invalid assignment

```python
s4 = Shape(4, 7)

try:
    s4.x = 9
except AttributeError as err:
    print(f"X: {err}")

try:
    s4.y = 9
except AttributeError as err:
    print(f"Y: {err}")
```

## Validate translate

```python
s5 = Shape(4, 7)

print(f"{s5.x = }")
print(f"{s5.y = }")
print()
s5.translate(-8, 3)
print(f"{s5.x = }") # Expected: -4
print(f"{s5.y = }") # Expected: 10
print()
s5.translate(2, 0)
print(f"{s5.x = }") # Expected: -2
print(f"{s5.y = }") # Expected: 10
```
