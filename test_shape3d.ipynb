# Manual testing for Shape3D class

## Test inherited coordinate functionality

### Valid translate tests

```python
from shape3d import Shape3D

class Cube(Shape3D): # Temporary class for testing
    @property
    def surface_area(self): ...

    @property
    def volume(self): ...

c1 = Cube(4, 1)
print("Before translate:")
print(f"{c1.x = }") # 4
print(f"{c1.y = }") # 1
print(f"{c1.z = }") # 0

c1.translate(2, 1)
print(f"\nAfter translate without moving z:")
print(f"{c1.x = }") # 6
print(f"{c1.y = }") # 2
print(f"{c1.z = }") # 0

c1.translate(4, 3, -3)
print(f"\nAfter translate with moving z:")
print(f"{c1.x = }") # 10
print(f"{c1.y = }") # 5
print(f"{c1.z = }") # -3
```

### Invalid translate tests

```python
c2 = Cube(4, 1, 5)

try:
    c2.translate(2, 1, "...")
except TypeError as err:
    print(err)
    print(f"{c2.x = }")
    print(f"{c2.y = }")
    print(f"{c2.z = }")

try:
    c2.translate("...", 1, 4)
except TypeError as err:
    print(err)
    print(f"{c2.x = }")
    print(f"{c2.y = }")
    print(f"{c2.z = }")

try:
    c2.translate(2, "...", 4)
except TypeError as err:
    print(err)
    print(f"{c2.x = }")
    print(f"{c2.y = }")
    print(f"{c2.z = }")
```
