## Operator overload tests with separate subclasses

```python
from sphere import Sphere
from cube import Cube

c1 = Cube(4)
sph1 = Sphere(5)

print(f"{c1.volume = }")
print(f"{sph1.volume = }")
print()
print(f"{c1 == sph1 = }") # Expected: False
print(f"{c1 < sph1 = }") # Expected: True
print(f"{c1 > sph1 = }") # Expected: False
print(f"{c1 <= sph1 = }") # Expected: True
print(f"{c1 >= sph1 = }") # Expected: False
```
