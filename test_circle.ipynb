## translate() test

```python
c1 = Circle(5, 0, 9)

c1.translate(3, 1) # Expected new positions: 3, 10
print(c1)

c1.translate(-1, 7) # Expected new positions: 2, 17
print(c1)
```

## Dunder tests

### \_\_eq\_\_ tests

```python
c1 = Circle(5, 2, 1)
c2 = Circle(2, 1, 9)
c3 = Circle(2, 4, 9)

print(c1 == c2) # Expected: False
print(c2 == c3) # Expected: True
print(c1 == "...") # Expected: False
print(c1 == 5) # Expected: False
print(c1 == True) # Expected: False
```

### \_\_lt\_\_ tests

```python
print(f"{c1 < c3 = }") # Expected: False
print(f"{c2 < c1 = }") # Expected: True
print(f"{c2 < c3 = }") # Expected: False
```

### \_\_gt\_\_ tests

```python
print(f"{c1 > c2 = }") # Expected: True
print(f"{c2 > c1 = }") # Expected: False
print(f"{c2 > c3 = }") # Expected: False
```

### \_\_ge\_\_ tests

```python
print(f"{c1 >= c2 = }") # Expected: True
print(f"{c2 >= c3 = }") # Expected: True
print(f"{c2 >= c1 = }") # Expected: False
```

### \_\_le\_\_ tests

```python
print(f"{c1 <= c2 = }") # Expected: False
print(f"{c2 <= c3 = }") # Expected: True
print(f"{c2 <= c1 = }") # Expected: True
```
