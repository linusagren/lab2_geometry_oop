# Manual testing for Circle class

## Test valid init

```python
from circle import Circle

c1 = Circle(4, 2, 8)
c1
```

## Initialization: Invalid init value tests

```python
try:
    c1 = Circle(0) # Expected: ValueError
except ValueError as err:
    print(f"Circle(0): {err}")

try:
    c1 = Circle(-5) # Expected: ValueError
except ValueError as err:
    print(f"Circle(-5): {err}")

try:
    c1 = Circle("...") # Expected: TypeError
except TypeError as err:
    print(f"Circle('...'): {err}")

try:
    c1 = Circle(True) # Expected: TypeError
except TypeError as err:
    print(f"Circle(True): {err}")
```

## Initialization: Valid values

```python
c1 = Circle(2.53)
print(f"{c1 = }")

c2 = Circle(6)
print(f"{c2 = }")
```

## Assignment: Invalid values

### Zero-values

```python
try:
    c1.radius = 0 # Expected: ValueError
except ValueError as err:
    print(f"c1.radius = 0: {err}")

try:
    c1.radius = -5 # Expected ValueError
except ValueError as err:
    print(f"c1.radius = -5: {err}")

try:
    c1.radius = "..." # Expected: TypeError
except TypeError as err:
    print(f"c1.radius = '...': {err}")

try:
    c1.radius = True # Expected: TypeError
except TypeError as err:
    print(f"c1.radius = True: {err}")
```

## Dunder tests

### \_\_eq\_\_ tests

```python
c1 = Circle(5, 2, 1)
c2 = Circle(2, 1, 9)
c3 = Circle(2, 4, 9)

print(f"{c1 == c2 = }") # Expected: False
print(f"{c2 == c3 = }") # Expected: True
print(f"{c1 == "..." = }") # Expected: False
print(f"{c1 == 5 = }") # Expected: False
print(f"{c1 == True = }") # Expected: False
```

### \_\_lt\_\_ tests

```python
print(f"{c1 < c3 = }") # Expected: False
print(f"{c2 < c1 = }") # Expected: True
print(f"{c2 < c3 = }") # Expected: False
```

### \_\_gt\_\_ tests

```python
print(f"{c1 > c2 = }") # Expected: True
print(f"{c2 > c1 = }") # Expected: False
print(f"{c2 > c3 = }") # Expected: False
```

### \_\_ge\_\_ tests

```python
print(f"{c1 >= c2 = }") # Expected: True
print(f"{c2 >= c3 = }") # Expected: True
print(f"{c2 >= c1 = }") # Expected: False
```

### \_\_le\_\_ tests

```python
print(f"{c1 <= c2 = }") # Expected: False
print(f"{c2 <= c3 = }") # Expected: True
print(f"{c2 <= c1 = }") # Expected: True
```
