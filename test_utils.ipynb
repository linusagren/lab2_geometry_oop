## Rare NaN edge-case
### Initial test

```python
value = float("nan")

try:
    if (value <= 0) == False and (value > 0) == False:
        raise ValueError(f"Value cannot be NaN.")
except ValueError as err:
    print(err)
```

### Check how NaN compares to itself

```python
NaN = float("nan")

print(f"{NaN = }")
print(f"{NaN == NaN = }")
print(f"{NaN != NaN = }")
```

### Simplified comparison

```python
try:
    if value != value:
        raise ValueError("Value cannot be NaN.")
except ValueError as err:
    print(err)
```
